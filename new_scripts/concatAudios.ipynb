```python
import os, glob, soundfile as sf

# ── 1. Configuration ────────────────────────────────────────────────
folder_path = '../Data/nestingbox1/20240927_204919'
output_file = os.path.join(folder_path, 'combined_audio.w64')
# ── 2. Discover inputs ─────────────────────────────────────────────
wav_paths = sorted(glob.glob(os.path.join(folder_path, '*.wav')))
print(f"🔍 Found {len(wav_paths)} .wav files in {folder_path!r}:")
for i, p in enumerate(wav_paths, 1):
    info = sf.info(p)
    dur_hr = info.frames / info.samplerate / 3600
    print(f"   {i:2d}. {os.path.basename(p)} — {dur_hr:.2f} h ({info.frames} frames @ {info.samplerate} Hz)")

if not wav_paths:
    raise RuntimeError("No .wav files found – check your folder_path!")

# ── 3. Read header from the first file ─────────────────────────────
with sf.SoundFile(wav_paths[0], 'r') as src:
    samplerate = src.samplerate
    channels   = src.channels
    subtype    = src.subtype

# ── 4. Stream‐write into one big file ──────────────────────────────
print("\n🛠  Concatenating…")
with sf.SoundFile(
    output_file, mode='w',
    samplerate=samplerate,
    channels=channels,
    subtype=subtype,
    format='W64'             # ← or 'RF64' if your libsndfile build supports it
) as dst:
    for path in wav_paths:
        with sf.SoundFile(path, 'r') as src:
            while True:
                block = src.read(1024*1024, dtype='float64')
                if not len(block):
                    break
                dst.write(block)

print("✅ Written large-file WAV to", output_file)

# ── 5. Report final duration ───────────────────────────────────────
combined_info = sf.info(output_file)
total_hr = combined_info.frames / combined_info.samplerate / 3600
print(f"\n✅ Written {output_file!r} — {total_hr:.2f} h ({combined_info.frames} frames)")
```
